```python
X=df.drop('Price',axis=1)
y=df.Price
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.8,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
scaler = StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
poly = PolynomialFeatures(degree=4,include_bias=False )

  
X_train_poly = poly.fit_transform(X_train_scaled) 
X_test_poly = poly.transform(X_test_scaled) 
  
print("Original Features :", X_train.shape[1])
print("Polynomial Features :", X_train_poly.shape[1]) 
```

```python
ridge=Ridge(alpha=1.0)
model=LinearRegression
```

```python
linear = LinearRegression() 
  
linear.fit(X_train_poly, y_train) 
  
train_pred = linear.predict(X_train_poly)
test_pred = linear.predict(X_test_poly) 
  
train_r2 = r2_score(y_train, train_pred) 
test_r2 = r2_score(y_test, test_pred) 
  
print("LINEAR REGRESSION")
print("Training R2 :", round(train_r2,4)) 
print("Testing R2 :", round(test_r2,4)) 
  
```

```python
ridge.fit(X_train_scaled, y_train)
ridge_pred = ridge.predict(X_test_scaled)
ridge_mae=mean_absolute_error(y_test,ridge_pred)
ridge_mse=mean_squared_error(y_test,ridge_pred)
ridge_rmse=ridge_mse** 0.5
ridge_r2=r2_score(y_test,ridge_pred)
n=X_test.shape[0]
p=X_test.shape[1]
ridge_adj_r2=1-((1-ridge_r2)*(n-1)/(n-p-1))
ridge_train = r2_score(y_train, train_pred)
ridge_test = r2_score(y_test, test_pred) 
print("--------RIDGE REGRESSION---------")
print("MAE:",ridge_mae)
print("MSE:",ridge_mse)
print("RMSE:",ridge_rmse)
print("R2 SCore:",ridge_r2)
print("Adjusted R2:",ridge_adj_r2)
print("Training R2 :", round(ridge_train,4))
print("Testing R2 :", round(ridge_test,4)) 
print("\nRidge Coefficients:")
for feature, coef in zip(
    X.columns,
    ridge.coef_
):
    print(feature,":",coef)
```

```python
lasso=Lasso(alpha=1000)
lasso.fit(X_train_scaled, y_train)
lasso_pred=lasso.predict(X_test_scaled)
lasso_mae =mean_absolute_error(y_test,lasso_pred)
lasso_mse =mean_squared_error(y_test,lasso_pred)
lasso_rmse=lasso_mse ** 0.5
lasso_r2=r2_score(y_test,lasso_pred)
lasso_adj_r2=1-((1-lasso_r2)*(n-1)/(n-p-1))
lasso_train = r2_score(y_train, train_pred) 
lasso_test = r2_score(y_test, test_pred) 
print("-----LASSO REGRESSION--------")
print("MAE:",lasso_mae)
print("MSE:",lasso_mse)
print("RMSE:",lasso_rmse)
print("R2 Score:",lasso_r2)
print("Adjusted R2:",lasso_adj_r2)
print("Training R2 :", round(lasso_train,4)) 
print("Testing R2 :", round(lasso_test,4)) 
print("n\Lasso Coefficients:")
for feature, coef in zip(
    X.columns,
    lasso.coef_
):
    print(feature,":",coef)
```

```python
elastic = ElasticNet(alpha=0.1,l1_ratio=0.5)
elastic.fit(X_train_scaled,y_train)
elastic_pred=elastic.predict(X_test_scaled)
elastic_mae=mean_absolute_error(y_test,elastic_pred)
elastic_mse=mean_squared_error(y_test,elastic_pred)
elastic_rmse=elastic_mse ** 0.5
elastic_r2=r2_score(y_test,elastic_pred)
elastic_adj_r2=1-((1-elastic_r2)*(n-1)/(n-p-1))
elastic_train = r2_score(y_train, train_pred)
elastic_test = r2_score(y_test, test_pred) 
print("-------ELASTIC NET REGRESSION-------")
print("MAE:",elastic_mae)
print("MSE:",elastic_mse)
print("RMSE:",elastic_rmse)
print("R2:",elastic_r2)
print("Adjusted R2:",elastic_adj_r2)
print("Training R2 :", round(elastic_train,4)) 
print("Testing R2 :", round(elastic_test,4)) 
  


print("n\ElasticNet Coefficients:")
for feature ,coef in zip(
    X.columns,
    elastic.coef_
):
    print(feature,":",coef)
```

```python
results=pd.DataFrame({"Model":["Ridge","Lasso","ElasticNet"],"MAE":[ridge_mae,lasso_mae,elastic_mae],"MSE":[ridge_mse,lasso_mse,elastic_mse],"RMSE":[ridge_rmse,lasso_rmse,elastic_rmse],"R2":[ridge_r2,lasso_r2,elastic_r2],"Adjusted R2":[ridge_adj_r2,lasso_adj_r2,elastic_adj_r2]
                     }) 
print(results)
```
